## Ejercicio 5 — Funciones propias para el cálculo de importes
Defino calcular_importe con el descuento como parámetro opcional y resumen_pedido, que devuelve una tupla de tres valores.
Llamo a resumen_pedido con argumentos por palabra clave y desempaqueto el resultado.
Por último aplico calcular_importe a los cinco pedidos y uso max(), min(), sorted() y len().

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """Devuelve el importe neto (con descuento) redondeado a dos decimales."""
    bruto = unidades * precio_unitario
    return round(bruto * (1 - descuento_pct / 100), 2)
def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """Devuelve una tupla (importe_bruto, ahorro, importe_neto)."""
    bruto = round(unidades * precio_unitario, 2)
    neto = calcular_importe(unidades, precio_unitario, descuento_pct)
    ahorro = round(bruto - neto, 2)
    return (bruto, ahorro, neto)
bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print("Bruto:", bruto, "| Ahorro:", ahorro, "| Neto:", neto)
help(calcular_importe)
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]
importes = [calcular_importe(*pedido) for pedido in pedidos]
print("Importes:", importes)
print("Máximo:", max(importes))
print("Mínimo:", min(importes))
print("Ordenados:", sorted(importes))
print("Nº de pedidos:", len(importes))
print("Comprobación:", calcular_importe(3, 59.90, 10))
```

## Ejercicio 8 — Del cálculo con listas al cálculo vectorizado
He intentado multiplicar dos listas (da error), las he convertido a arrays de NumPy y he calculado los importes, su suma y el recargo del 4 % con operaciones vectorizadas.
Después he creado un array mixto y he comparado el tiempo del bucle con el de la versión vectorizada usando %timeit.

```python
unidades = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios  = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]
try:
    print(unidades * precios)
except TypeError as e:
    print("TypeError:", e)
np_unidades = np.array(unidades)
np_precios = np.array(precios)
print("unidades:", np_unidades.dtype, np_unidades.shape)
print("precios:", np_precios.dtype, np_precios.shape)
importes = np_unidades * np_precios
print("Importes:", importes)
print("Suma total:", round(importes.sum(), 2))
print("Productos con precio > 100:", (np_precios > 100).sum())
importes_recargo = importes * 1.04
print("Con recargo:", importes_recargo.round(2))
mixto = np.array([1, "dos", 3.0, True])
print(mixto, mixto.dtype)
def importes_con_bucle():
    resultado = []
    for i in range(len(unidades)):
        resultado.append(unidades[i] * precios[i])
    return resultado
%timeit importes_con_bucle()
%timeit np_unidades * np_precios
```

## Ejercicio 11 — Estadística descriptiva sobre telemetría simulada
Genero la telemetría con un generador con semilla, calculo los estadísticos y percentiles, apilo ambas series en una matriz, calculo la correlación y cuento las lecturas que superan el umbral de media + 2 desviaciones.

```python
rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

media = temperatura.mean()
desviacion = temperatura.std()
print("Media   :", round(media, 2))
print("Mediana :", round(np.median(temperatura), 2))
print("Desv.   :", round(desviacion, 2))
print("Mínimo  :", round(temperatura.min(), 2))
print("Máximo  :", round(temperatura.max(), 2))

print("Percentiles 25/50/75/95:", np.percentile(vibracion, [25, 50, 75, 95]).round(2))

datos = np.column_stack((temperatura, vibracion))
print("shape:", datos.shape)

print(np.corrcoef(temperatura, vibracion).round(3))

umbral = media + 2 * desviacion
superan = (temperatura > umbral).sum()
print("Umbral:", round(umbral, 2), "| Lecturas por encima:", superan,
      "| Porcentaje:", round(superan / len(temperatura) * 100, 2), "%")
```

## Ejercicio 15 — Columnas calculadas, agregación y exportación

```python
def nivel_riesgo(temperatura):
    """Clasifica una temperatura en CRÍTICO, ALTO, MEDIO o BAJO."""
    if temperatura > 85:
        return "CRÍTICO"
    elif temperatura > 76:
        return "ALTO"
    elif temperatura > 70:
        return "MEDIO"
    else:
        return "BAJO"

for valor in [92.5, 80.0, 72.3, 64.1]:
    print(valor, "->", nivel_riesgo(valor))

sensores["riesgo"] = sensores["temperatura_c"].apply(nivel_riesgo)
print(sensores[["id_maquina", "temperatura_c", "riesgo"]].head())

print(sensores["riesgo"].value_counts())

criticas = sensores[sensores["riesgo"] == "CRÍTICO"]
print(criticas["id_maquina"].value_counts())

print(sensores["estado"].value_counts())
```

```python
for etiqueta, fila in ventas.head(5).iterrows():
    print(fila["region"], "|", fila["producto"], "|", fila["importe_neto"], "€")

ventas["codigo_region"] = ventas["region"].str[:3].str.upper()

ventas["segmento_ticket"] = ventas["importe_neto"].apply(clasificar_ticket)

print(ventas["segmento_ticket"].value_counts())

sensores["eficiencia"] = (sensores["unidades_producidas"] / sensores["temperatura_c"]).round(3)
print(sensores[["id_maquina", "eficiencia"]].head())

resumen_por_region = {}

for region in ventas["region"].unique():
    filtrado = ventas[ventas["region"] == region]
    resumen_por_region[region] = round(filtrado["importe_neto"].sum(), 2)

print(resumen_por_region)

con_groupby = ventas.groupby("region")["importe_neto"].sum().round(2)
print(con_groupby)
print("Coinciden:", all(resumen_por_region[r] == con_groupby[r] for r in resumen_por_region))
```
