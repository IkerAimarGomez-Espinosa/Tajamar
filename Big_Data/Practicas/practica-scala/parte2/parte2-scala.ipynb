# Ejercicio 5 — Comparación de dos manos
Escribes la función que reciba ambos valores de las manos a comparar y para que responda cual es el numero más alto que ha recibido.
Dentro de la función comparas si la mano 1 es mayor a la segunda, si lo es que la devuelva, y si no que devuelva la segunda.

```scala
def maxHand(hand1: Int, hand2: Int): Int = {
  if (hand1 > hand2) {
    hand1
  } else {
    hand2
  }
}
println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

Cuando ambas tienen el mismo valor, devuelve el valor de la hand2, ya que el requisito para que devuelva hand1 es unicamente si esta es mayor que hand2.
No diría que es estrictamente necesario modificar la solución, ya que en caso de empate, aunque devuelva el valor b, es el mismo que el valor a, y por lo tanto cumple la función que se espera en el ejercicio, aunque creo que devolver un string indicando la mano victoriosa y pudiendo indicar si hubo empate, sería una devolución mas clara para los que ejecuten el codigo.

```scala
def maxHand(hand1: Int, hand2: Int): String = {
  if (hand1 > hand2) {
    "Mano 1 ha sido mayor que Mano 2"
  } else if (hand2 > hand1) {
    "Mano 2 ha sido mayor que Mano 1"
  } else {
      "Ambas manos tienen el mismo valor"
  }
}
println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

# Ejercicio 6 — Decidir el ganador de una partida
Defines la def ganador con varios condicionales dependiendo de las posibles comparaciones entre las dos manos que tienes que recibir.
Dentro de la primera comprobación, para ver si ambos jugadores han busteado, llamas dos veces a la función bust creada hace unas celdas.
Dentro de los dos siguientes else if, compruebas si cada uno de los dos jugadroes a busteado unicamente el.
Una vez has comprobado que ninguno de los dos jugadores ha busteado comprobamos que puntuación es mayor y la imprimimos.

```scala
def ganador(hand1: Int, hand2: Int): Int = {
  if (bust(hand1) && bust(hand2)) {
    0
  } else if (bust(hand1)) {
    hand2
  } else if (bust(hand2)) {
    hand1
  } else if (hand1 > hand2) {
    hand1
  } else {
    hand2
  }
}
println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

- Para 26, 20 se ha usado else if (bust(hand1))
- Para 18, 22 se ha usado else if (bust(hand2))
- Para 24, 25 se ha usado if (bust(hand1) && bust(hand2))
- Para 17, 19 se ha usado el else final
- Para 21, 20 se ha usado else if (hand1 >= hand2)

# Ejercicio 7 — Arrays y mutabilidad
Creamos el Array indicado y lo imprimos en forma de String.
Sustituimos el nombre de la primera posicion del array (la posicion 0) y lo volvemos a imprimir.
Intentamos sustituir la segunda posición del array por un numero.

```scala
val jugadoresArray = Array("Alex", "Chen", "Marta")
println(jugadoresArray.mkString(", "))
jugadoresArray(0) = "Sindhu"
println(jugadoresArray.mkString(", "))
jugadoresArray(1) = 500
```

Los valores dentro de el array pueden cambiar aunque sea un val por que inmutabilidad del val afecta al array como "caja" de lo que tiene dentro, y los valores del interior no se ven afectados por esta restricción.
Aunque tu no indiques en ningún momento que el array solo debería de poder contener Strings, Scala al solop ver strings en su creación original, lo interpreta como que será un array unicamente de strings y no deja introducir ningun otro tipo de valor.
La diferencia entre reasignar la variable y modificar los elementos del array es que reasignar la variable aunque el obejtivo sea solo cambiar el contenido, Scala lo interpreta como "cambiar la caja completamente" y eso val no lo permite, mientras que ir cambiando el contenido no supone ningún problema siempre que no haya una restricción de tipo.

# Ejercicio 9 — Recorrer un Array con while
Copiamos el array y creamos el contador i.
Creamos un while que indica que imprimira todas las posiciones del array manos hasta que i deje de ser menos a la longitud del array.
Dentro de cada iteración vemos el valor de la mano y si cumple la función bust, además de aumentar en uno el valor de i.

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0
while (i < manos.length) {
  println("Valor mano: " + manos(i) + " Bust: " + bust(manos(i)))
  i += 1
}
```

# Ejercicio 15 — Programa integrado: torneo de Twenty-One

```scala
val jugadoresTorneo = List("Alex", "Chen", "Marta", "Sindhu")
val manosTorneo = Array(18, 24, 21, 20)
def mejorMano(manos: Array[Int]): Int = {
  var i = 0
  var mejor = 0
  while (i < manos.length) {
    if (!bust(manos(i)) && manos(i) > mejor) {
      mejor = manos(i)
    }
    i = i + 1
  }
  mejor
}
var i = 0
while (i < manosTorneo.length) {
  if (bust(manosTorneo(i))) {
    println(jugadoresTorneo(i) + ": " + manosTorneo(i) + " - Se ha pasado")
  } else {
    println(jugadoresTorneo(i) + ": " + manosTorneo(i) + " - Puntuación válida")
  }
  i += 1
}
println("Mejor puntuación válida: " + mejorMano(manosTorneo))
```
